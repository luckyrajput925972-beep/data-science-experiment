```python
# Q29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary.

sales = pd.Series(
    [300, 700, 450, 900],
    index=['Pen', 'Bag', 'Book', 'Laptop']
)

print(sales.values)
print(sales.index)
print(sales['Bag'])
print(sales[sales > 500])
print('Book' in sales.index)
print(sales.to_dict())
```

```python
# Q30. Analyse the following code and answer all parts: (a) What is the index order of s? (b) Which label contains NaN? (c) What will the Boolean filter s[s > 20] return? (d) What will pd.notnull(s) indicate for each label? (e) What are the Series name and index name after the final two statements?

data = {'A': 15, 'B': 35, 'C': 25}

s = pd.Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'
s.index.name = 'Category'

print(s)
print(s.index)
print(s[s > 20])
print(pd.notnull(s))
print(s.name)
print(s.index.name)
```
