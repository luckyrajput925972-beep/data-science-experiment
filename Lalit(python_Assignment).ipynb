```python
# Q7.1: Create my_abs() and test positive, negative and zero.
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-7))
print(my_abs(0))
```

```python
# Q7.2: Handle a string input safely.
def my_abs(val):
    if not isinstance(val, (int, float)):
        return "Please enter a number"

    if val < 0:
        return 0 - val
    return val

print(my_abs("Hi"))
print(my_abs(-7))
```

```python
# Q7.3: Create print_abs() and check the returned value.
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)

# x is None because the function does not return a value.
```

```python
# Q7.4: Check whether inc_val() changes the original x.
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)

# Output is 7 because val is reassigned locally.
```

```python
# Q7.5: Swap local values without changing the originals.
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print(val1, val2)

x = 6
y = 3

swap_message(x, y)
print(x, y)
```

```python
# Q8.2: Create and print a global counter.
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3: Try incrementing counter without the global keyword.
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)

# UnboundLocalError occurs because Python treats counter as local here.
```

```python
# Q8.4: Increment the global counter correctly.
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
