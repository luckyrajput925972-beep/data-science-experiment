```python
# Q1. Create a NumPy array containing the values 10, 20, 30, 40, and 50.

import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

```python
# Q2. Create a NumPy array of five zeros.

arr = np.zeros(5)
print(arr)
```

```python
# Q3. Create a NumPy array of five ones.

arr = np.ones(5)
print(arr)
```

```python
# Q4. Create an array of integers from 1 to 10.

arr = np.arange(1, 11)
print(arr)
```

```python
# Q5. Create an array of even numbers from 2 to 20.

arr = np.arange(2, 21, 2)
print(arr)
```

```python
# Q6. Create an array of five equally spaced values between 0 and 10.

arr = np.linspace(0, 10, 5)
print(arr)
```

```python
# Q7. Create a 2D NumPy array with two rows and three columns.

arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
```

```python
# Q8. Create a 3 × 3 identity matrix.

arr = np.eye(3)
print(arr)
```

```python
# Q9. Create a NumPy array containing the values 5, 10, 15, 20, and 25, then display the first element.

arr = np.array([5, 10, 15, 20, 25])
print(arr[0])
```

```python
# Q10. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then display the last element.

arr = np.array([10, 20, 30, 40, 50])
print(arr[-1])
```

```python
# Q11. Create a NumPy array containing the values 1, 2, 3, 4, 5, and 6, then display its shape, size, and data type.

arr = np.array([1, 2, 3, 4, 5, 6])

print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
# Q12. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then display the elements from index 1 to index 3.

arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4])
```

```python
# Q13. Create a NumPy array containing the values 1, 2, 3, 4, and 5, then reverse the array.

arr = np.array([1, 2, 3, 4, 5])
print(arr[::-1])
```

```python
# Q14. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then reshape it into a 5 × 1 array.

arr = np.array([10, 20, 30, 40, 50])
print(arr.reshape(5, 1))
```

```python
# Q15. Create a 3 × 3 array containing the values 1 to 9, then display the second row.

arr = np.arange(1, 10).reshape(3, 3)
print(arr[1])
```

```python
# Q16. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then find its sum.

arr = np.array([10, 20, 30, 40, 50])
print(np.sum(arr))
```

```python
# Q17. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then find its mean.

arr = np.array([10, 20, 30, 40, 50])
print(np.mean(arr))
```

```python
# Q18. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then find its maximum and minimum values.

arr = np.array([10, 20, 30, 40, 50])

print(np.max(arr))
print(np.min(arr))
```

```python
# Q19. Create a NumPy array containing the values 5, 10, 15, 20, and 25, then calculate the cumulative sum.

arr = np.array([5, 10, 15, 20, 25])
print(np.cumsum(arr))
```

```python
# Q20. Create two NumPy arrays and perform addition, subtraction, multiplication, and division.

arr1 = np.array([10, 20, 30])
arr2 = np.array([2, 4, 5])

print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

```python
# Q21. Create a NumPy array containing the values 10, 25, 30, 45, and 50, then display elements greater than 30.

arr = np.array([10, 25, 30, 45, 50])
print(arr[arr > 30])
```

```python
# Q22. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then replace all elements greater than 30 with 0.

arr = np.array([10, 20, 30, 40, 50])
arr[arr > 30] = 0
print(arr)
```

```python
# Q23. Create a NumPy array containing the values 10, 20, 30, 40, and 50, then sort the array in descending order.

arr = np.array([10, 20, 30, 40, 50])
print(np.sort(arr)[::-1])
```

```python
# Q24. Create a NumPy array containing duplicate values and find its unique elements.

arr = np.array([1, 2, 2, 3, 4, 4, 5])
print(np.unique(arr))
```

```python
# Q25. Create a NumPy array containing the values 1, 2, 3, 4, and 5, then find the square of each element.

arr = np.array([1, 2, 3, 4, 5])
print(np.square(arr))
```

```python
# Q26. Create a NumPy array containing the values 1, 4, 9, 16, and 25, then calculate the square root of each element.

arr = np.array([1, 4, 9, 16, 25])
print(np.sqrt(arr))
```

```python
# Q27. Create a NumPy array containing the values 35, 62, 48, 81, and 55, then use np.where() to display Fast for speeds of at least 60 and Normal otherwise.

speed = np.array([35, 62, 48, 81, 55])

result = np.where(speed >= 60, "Fast", "Normal")
print(result)
```

```python
# Q28. Create a 2D NumPy array containing the values 1 to 9 and calculate the sum of each row and each column.

arr = np.arange(1, 10).reshape(3, 3)

print(np.sum(arr, axis=1))
print(np.sum(arr, axis=0))
```
